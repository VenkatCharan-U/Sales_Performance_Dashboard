```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("../DataSet/Superstore.xls")
```

```python
# Business Question 1

## What are the total sales generated by the company?

Understanding total sales provides an overview of the company's revenue performance.
```

```python
total_sales = df["Sales"].sum()
print(f"Total Sales : ${total_sales:,.2f}")
```

```python
total_profit = df["Profit"].sum()

print(f"Total Profit : ${total_profit:,.2f}")
```

```python
total_orders = df["Order ID"].nunique()
print(f"Total Number of Orders : {total_orders}")
```

```python
average_order_value = total_sales/total_orders
print(f"Average Order Value : ${average_order_value:,.2f}")
```

```python
total_customers = df["Customer ID"].nunique()
print(f"The number of Customers : {total_customers}")
```

```python
total_products = df["Product ID"].nunique()
print(f"The number of Products : {total_products}")
```

```python
## Business Observation

The company generated a total revenue of **$2.33 million** from **5,111 customer orders**, resulting in an average order value of **$455.20**.

The company served **804 unique customers** and sold **1,862 unique products**, indicating a broad customer base and diverse product catalog.

Overall profitability reached **$292,296.81**, suggesting that the business is profitable, although further analysis is needed to identify which products, customers, or regions contribute most to this profit.
```

```python
profit_margin = (total_profit / total_sales) * 100
print(f"Prodit Margin : {profit_margin:,.2f}")
```

```python
average_profit_per_order = total_profit/total_orders
print(f"Average Profit Per Order : ${average_profit_per_order:,.2f}")
```

```python
average_sales_per_customer = total_sales/total_customers
print(f"Average Sales per Customer : ${average_sales_per_customer:,.2f}")
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
        .sum()
        .sort_values(ascending=False)
)
category_sales.to_frame(name="Total Sales")
```

```python
plt.figure(figsize=(8,5))

category_sales.plot(
    kind="bar",
    color=["royalblue","orange","green"]
)

plt.title("Sales by Category",fontsize=16)

plt.xlabel("Category",fontsize=12)

plt.ylabel("Sales ($)",fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y",linestyle="--",alpha=0.4)

plt.show()
```

```python
category_profit = (
    df.groupby("Category")["Profit"]
        .sum()
        .sort_values(ascending=False)
)
category_profit.to_frame(name="Total Profit")
```

```python
plt.figure(figsize=(8,5))

category_profit.plot(
    kind="bar",
    color=["royalblue","green","red"]
)

plt.title("Profit by Category",fontsize=16)

plt.xlabel("Category",fontsize=12)

plt.ylabel("Profit ($)",fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

```python
## Business Insight

Technology is the highest-performing category in terms of both sales and profit.

Although Furniture generates the second-highest sales, it contributes the lowest overall profit, indicating lower profit margins.

Office Supplies generates lower sales than Furniture but produces significantly higher profits.

## Recommendation

- Continue investing in Technology products because they generate both strong revenue and high profitability.
- Investigate Furniture pricing, discounts, and operational costs to improve profitability.
- Promote Office Supplies through targeted marketing campaigns, as it delivers strong profit relative to its sales.
```

```python
subcategory_sales = (
    df.groupby("Sub-Category")["Sales"]
        .sum()
        .sort_values(ascending=False)
)

subcategory_sales.to_frame(name="Total Sales by Sub Category")
```

```python
ax = subcategory_sales.plot(
    kind="bar",
    figsize=(12,6),
    color="steelblue"
)

plt.title("Sales by Sub Category")
plt.xlabel("Sub Category")
plt.ylabel("Sales ($)")

for bar in ax.patches:
    ax.annotate(
        f'{bar.get_height():,.0f}',
        (bar.get_x() + bar.get_width()/2, bar.get_height()),
        ha='center',
        va='bottom',
        fontsize=8,
        rotation=0
    )

plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
            .sum()
            .sort_values(ascending=False)
)

subcategory_profit.to_frame(name="Total Profit by Sub-Category")
```

```python
ax = subcategory_profit.plot(
    kind="bar",
    figsize=(12,6),
    color="steelblue"
)

plt.title("Profit by Sub Category")
plt.xlabel("Sub Category")
plt.ylabel("Profit ($)")

for bar in ax.patches:
    ax.annotate(
        f'{bar.get_height():,.0f}',
        (bar.get_x() + bar.get_width()/2, bar.get_height()),
        ha='center',
        va='bottom',
        fontsize=8,
        rotation=0
    )

plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
## Business Insight

- Chairs generated the highest sales revenue among all sub-categories, indicating strong customer demand. It also contributes healthy profits, making it one of the company's most valuable product lines.

- Copiers generated the highest total profit despite lower sales than Chairs and Phones. This suggests Copiers have a significantly higher profit margin and should be considered a premium product category.

- Tables generated relatively high sales but recorded the largest overall loss, indicating potential issues such as excessive discounts, high procurement costs, or low pricing.

- Bookcases and Supplies also generated negative profits, making them candidates for further business investigation.

## Recommendation

1. Increase investment and marketing for Chairs, Phones, and Copiers since they are strong contributors to both revenue and profitability.

2. Analyze pricing strategies and discount policies for Tables, Bookcases, and Supplies to identify the causes of financial losses.

3. Review supplier costs and operational expenses for loss-making products before considering discontinuation.

4. Avoid immediately removing loss-making products. Instead, perform a root-cause analysis to determine whether pricing, discounts, costs, or customer demand are responsible for the losses.
```

```python
# Business Question 5

## Which region generates the highest sales?

Understanding regional sales helps management allocate resources, marketing budgets, and inventory more effectively.
```

```python
region_sales = (
    df.groupby("Region")["Sales"]
        .sum()
        .sort_values(ascending=False)
    
)

region_sales.to_frame(name="Total Sales")
```

```python
plt.figure(figsize=(8,5))

region_sales.plot(
    kind="bar",
    color="cornflowerblue"
)

plt.title("Sales by Region",fontsize=16)

plt.xlabel("Region",fontsize=12)

plt.ylabel("Sales ($)",fontsize=12)

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
## Business Insight

- The West region generated the highest sales revenue ($739,813.61), making it the company's strongest-performing market.

- The East region closely follows with total sales of $691,828.17, indicating another highly valuable business region.

- The Central region generated moderate sales, while the South region recorded the lowest sales among all four regions.

- The combined sales from the West and East regions account for a significant portion of the company's overall revenue, highlighting their importance to the business.

## Recommendation

- Continue investing in the West and East regions through marketing campaigns, inventory availability, and customer engagement initiatives.

- Investigate the South region to understand the reasons for its lower sales, such as customer demand, competition, or limited product availability.

- Allocate sales targets and promotional campaigns according to regional performance.
```

```python
# Business Question 7

## Which states generate the highest sales?

This analysis identifies the strongest-performing markets and helps management prioritize investments.
```

```python
## Business Insight

- California is the highest revenue-generating state with sales exceeding $457,000, significantly outperforming every other state.

- New York ranks second with over $310,000 in sales, making it another major revenue contributor.

- Texas, Washington, and Pennsylvania also generate substantial sales and represent important regional markets.

- The top ten states contribute a large share of total company sales, indicating that business performance is concentrated in a few high-performing states.

## Recommendation

- Continue investing in high-performing states such as California and New York through targeted marketing, customer retention programs, and inventory optimization.

- Analyze customer purchasing behavior in these states to identify best practices that can be replicated in other markets.

- Ensure sufficient inventory availability in these high-demand states to avoid lost sales opportunities.
```
