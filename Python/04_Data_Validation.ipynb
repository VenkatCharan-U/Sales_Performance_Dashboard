```python
# Dashboard Data Preparation
```

```python
dashboard_df = df.copy()

print("Dashboard Dataset Shape:", dashboard_df.shape)
```

```python
dashboard_df["Year"] = dashboard_df["Order Date"].dt.year
dashboard_df["Month"] = dashboard_df["Order Date"].dt.month
dashboard_df["Month Name"] = dashboard_df["Order Date"].dt.month_name()
dashboard_df["Quarter"] = dashboard_df["Order Date"].dt.quarter
```

```python
dashboard_df["Month Start"] = dashboard_df["Order Date"].dt.to_period("M").dt.to_timestamp()
```

```python
dashboard_df[["Order Date", "Year", "Month", "Month Name", "Quarter", "Month Start"]].head()
```

```python
dashboard_df["Profit Margin (%)"] = (
    dashboard_df["Profit"] / dashboard_df["Sales"]
) * 100
```

```python
dashboard_df[["Sales", "Profit", "Profit Margin (%)"]].head()
```

```python
dashboard_df["Shipping Days"] = (
    dashboard_df["Ship Date"] - dashboard_df["Order Date"]
).dt.days
```

```python
dashboard_df[["Order Date", "Ship Date", "Shipping Days"]].head()
```
